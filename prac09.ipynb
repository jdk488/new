```python
import seaborn as sns
import matplotlib.pyplot as plt
from collections import Counter
```

```python
df = sns.load_dataset('titanic')

print(df.head())
```

```python
print("Summary:\n", df.describe())

print("\nTransposed Summary:\n", df.describe().transpose())
```

```python
age_data = df['age'].dropna()

mean_age = age_data.mean()
median_age = age_data.median()
mode_age = age_data.mode()[0]

print(f"Mean Age: {mean_age}")
print(f"Median Age: {median_age}")
print(f"Mode Age: {mode_age}")
```

```python
plt.figure(figsize=(8, 4))

sns.boxplot(x="sex", y="age", hue="survived", data=df)

plt.title("Age Distribution by Gender & Survival")
plt.xlabel("Sex")
plt.ylabel("Age")

plt.show()
```

```python
plt.figure(figsize=(8, 4))

sns.violinplot(x='sex', y='age', hue='survived', data=df)

plt.title("Violin Plot: Age vs Gender & Survival")

plt.show()
```

```python
plt.figure(figsize=(6, 4))

sns.countplot(x="sex", hue="survived", data=df)

plt.title("Survival Count by Gender")

plt.show()
```

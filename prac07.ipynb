```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
import re
```

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')   # <-- IMPORTANT
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
text = "Hello everyone! I am first name last name. I am a loyal KSKA Git user all the way from Sangamwadi Empire. I have considerable knowledge about life, Python, C++, Java, Rust, Golang and Blockchain. For every smart contract, I lose one strand of my hair. In my free time, which by the way, I barely get, I like to swim."
```

```python
sentences = sent_tokenize(text)
print("Sentences:\n", sentences)
```

```python
words = word_tokenize(text)
print("\nWords:\n", words)
```

```python
clean_text = re.sub('[^a-zA-Z]', ' ', text)
print("\nAfter removing punctuation:\n", clean_text)
```

```python
stop_words = set(stopwords.words('english'))

tokens = word_tokenize(clean_text.lower())

filtered_words = [word for word in tokens if word not in stop_words]

print("\nTokens:\n", tokens)
print("\nFiltered Words:\n", filtered_words)
```

```python
ps = PorterStemmer()

words_list = ["write", "writing", "wrote", "writes", "reading", "reads"]

print("\nStemming:")
for w in words_list:
    print(w, "->", ps.stem(w))
```

```python
lemmatizer = WordNetLemmatizer()

text2 = "studies studying cries cry"
tokens2 = word_tokenize(text2)

print("\nLemmatization:")
for w in tokens2:
    print(w, "->", lemmatizer.lemmatize(w))
```

```python
from nltk.tokenize import word_tokenize

text3 = "Hello everyone this is a sample text Earth"
tokens3 = word_tokenize(text3)

pos_tags = nltk.pos_tag(tokens3)

print(pos_tags)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

sentence = "This is an example of term frequency meow meow meow"

def calculate_tfIdf(doc):
    vectorizer = TfidfVectorizer()
    tf_matrix = vectorizer.fit_transform(doc)
    feature_names = vectorizer.get_feature_names_out()
    return tf_matrix, feature_names

document = [sentence]

tf_matrix, feature_names = calculate_tfIdf(document)

print("\nTF-IDF Features:\n", feature_names)
print("TF-IDF Matrix:\n", tf_matrix.toarray())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('titanic')

print(df.head())
```

```python
plt.figure(figsize=(6, 4))

sns.histplot(df['age'], kde=True)

plt.title("Age Distribution")
plt.show()
```

```python
plt.figure(figsize=(5, 3))

sns.boxplot(x='class', y='age', data=df)

plt.title("Age vs Class")
plt.show()

print(df.describe().transpose())
```

```python
plt.figure(figsize=(5, 3))

sns.boxplot(x='class', y='age', data=df)

plt.title("Age vs Class")
plt.show()

print(df.describe().transpose())
```

```python
plt.figure(figsize=(5, 4))

sns.violinplot(x='class', y='age', data=df)

plt.title("Violin Plot - Age vs Class")
plt.show()
```

```python
plt.figure(figsize=(5, 4))

pq = sns.histplot(data=df, x='fare', bins=10, hue='survived')

for container in pq.containers:
    pq.bar_label(container)

plt.title("Fare Distribution by Survival")
plt.show()
```

```python
plt.figure(figsize=(5, 4))

sns.scatterplot(x='age', y='fare', data=df)

plt.title("Age vs Fare")
plt.show()
```

```python
plt.figure(figsize=(5, 4))

sns.scatterplot(x='age', y='fare', hue='survived', data=df)

plt.title("Age vs Fare (Survival)")
plt.show()
```
